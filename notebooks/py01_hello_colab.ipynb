## 二、print()：讓電腦說話
語法：`print(資料1, 資料2, ..., sep=' ', end='\n')`
- 多個資料用逗號 `,` 隔開，輸出時預設以**空格**分隔（`sep`）
- 每次輸出完預設會**換行**（`end`）

### 1. 輸出多個字串（逗號隔開，輸出會以空格分開）

```python
print("Hello", "Python")
```

### 2. 直接輸出數字或運算結果

```python
print(1 + 2)
print(999 + 88888888 + 88888 + 777777)
```

### 3. 結合字串與運算式

```python
print("1+2=", 1 + 2)
```

### 4. 字串可用 `+` 串接、`*` 重複

```python
print("Hello" + "Python!")
print("Go! " * 5)
```

### 5. 試試 sep 與 end

```python
print("2026", "11", "04", sep="/")
print("不換行", end="")
print("…接在後面")
```

## 三、input()：讓使用者輸入
語法：`變數 = input("提示字串")`

⚠️ `input()` 拿到的一定是**字串**；要計算時必須先用 `int()` 或 `float()` 轉換。

```python
name = input("請輸入您的名字：")
print("你好，", name)
```
