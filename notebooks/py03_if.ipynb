## 二、if：條件成立才執行
```python
if 條件式:
    敘述（要縮排！）
```
- 條件式後面一定要有冒號 `:`
- 區塊內要**縮排**（Colab 預設 2 格，常見 4 格，同一區塊要一致）

````python
age = 20
if age >= 18:
    print("你的年齡", age)
    print("成年")
````
