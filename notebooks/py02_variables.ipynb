## 一、變數命名規則
1. 第 1 個字必須是**英文字母或底線 `_`**，不能是數字
2. 之後可以是英文字母、數字、底線
3. **大小寫不同**：`Score` 和 `score` 是兩個變數
4. 不能用保留字（如 `if`、`for`、`print` 雖不是保留字但也別拿來當變數名稱！）

## 二、基本資料型態
| 型態 | 名稱 | 例子 |
|---|---|---|
| `int` | 整數 | `2`、`-15` |
| `float` | 浮點數 | `2.0`、`3.14` |
| `str` | 字串 | `"Hello"`、`'台北'` |
| `bool` | 布林值 | `True`、`False`（第一個字母大寫） |

用 `type()` 可以查型態：

```python
print(type(2), type(2.0), type("2"), type(True))
```

## 四、算術運算子
| 運算子 | 意義 | 例子 | 結果 |
|---|---|---|---|
| `+ - *` | 加減乘 | `7 * 2` | `14` |
| `/` | 除（結果是 float） | `7 / 2` | `3.5` |
| `//` | 整數除法（商） | `7 // 2` | `3` |
| `%` | 取餘數 | `7 % 2` | `1` |
| `**` | 次方 | `16 ** 0.5` | `4.0` |

優先順序：`**` → `* / // %` → `+ -`，有括號先算括號。

```python
# ✏️ 練習
# 先猜再執行：
print(7 / 2)
print(3 * 5 % 2 - 1)
print(7 // 2)
print(16 ** 0.5)
```
