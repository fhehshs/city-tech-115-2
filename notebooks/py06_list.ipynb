## 二、串列常用操作

```python
n = [10, 20, 30, 40]
print(n, len(n))          # 輸出、長度
for x in n:               # 逐一取出元素
    print(x, end=" ")
print()
for i in range(len(n)):   # 用索引取出
    print(i, n[i])
print(30 in n, 50 not in n)
print(n + [50, 60], [0] * 4)
print(n[1:3], n[:2], n[2:])  # 切片
n[0] = 99
n.append(70)              # 加在最後
print(n)
```

## 三、二維串列：先列後行
5 位同學 3 科成績 → 5 × 3 的二維串列，`c[i][j]` 是第 i 列第 j 行（都從 0 開始）。

```python
c = [[85, 90, 78],
     [72, 65, 80],
     [95, 88, 92],
     [60, 75, 70],
     [88, 94, 85]]
print(c[2])       # 第 3 位同學的成績
print(c[2][1])    # 第 3 位同學的第 2 科
for i in range(len(c)):
    for j in range(len(c[i])):
        print(c[i][j], end=" ")
    print()
```
