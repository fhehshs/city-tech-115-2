## 一、函式
`print()`、`input()`、`len()` 都是內建函式。也可以用 `def` 自己定義：
```python

    函式主體
    return 回傳值   # 沒有回傳值可以省略
```

````python
def sayhello():
    print("Hello Python!")

def add(a, b):
    return a + b

sayhello()
sayhello()
s = add(6, 10)
print(s)
````
