## 一、for 迴圈與 range()
| 寫法 | 產生的數 |
|---|---|
| `range(6)` | 0, 1, 2, 3, 4, 5 |
| `range(2, 6)` | 2, 3, 4, 5（不含 6） |
| `range(2, 15, 3)` | 2, 5, 8, 11, 14 |
| `range(10, 0, -2)` | 10, 8, 6, 4, 2 |

```python
for i in range(2, 15, 3):
    print(i, end=" ")
```

## 二、雙重迴圈
外迴圈控制「列」，內迴圈控制「每列印幾個」。`print(..., end="")` 讓輸出不換行。

```python
for i in range(4):
    for j in range(6):
        print("*", end="")
    print()
```

## 三、while 迴圈
條件成立就一直做，**一定要讓條件有機會變成 False**，否則會變無窮迴圈（按儲存格左邊的 ■ 可以停止）。

```python
i = 1
while i < 6:
    print(i)
    i = i + 1
```
